```python
#Funds
spy = yf.Ticker("SPY").history(start="2010-01-01")
vix = yf.Ticker("^VIX").history(start="2010-01-01")

#Adjust timezones so dates exactly match
spy.index = spy.index.tz_localize(None).normalize()
vix.index = vix.index.tz_localize(None).normalize()

```

```python
ret = spy['Close'].pct_change()
ann = np.sqrt(252) * 100 # converting daily std to annual
data = pd.DataFrame(index = spy.index)
data["vol_5d"] = ret.rolling(5).std() * ann # 5-day vol
data["vol_21d"] = ret.rolling(21).std() * ann # 21-day vol
data["vol_63d"] = ret.rolling(63).std() * ann # 63-day vol
data["vix"] = vix["Close"] # value at close on day t
data["fwd_vol_5d"] = ret.rolling(5).std().shift(-5) * ann # target: days t+1...t+5
data = data.dropna()
```

```python
#Plot 1 VIX vs SPY vol
fig, ax = plt.subplots(figsize=(13, 5))
ax.plot(data.index, data["vix"], lw = 1, label="VIX Market Forecast")
ax.plot(data.index, data["fwd_vol_5d"], lw = 1, alpha=0.7, label = "Actual next week SPY volatility")
ax.set_ylabel("Annualized Vol (%)")
ax.set_title("VIX vs SPY Vol")
ax.legend()
ax.grid(alpha=0.3)
plt.show()
```

**Observations:**
Vix sits above the actual volatility during calm periods.

During spikes actual volatility often exceeds VIX.

Actual volatility is more reactive.

```python
# Plot 2: seeing which predictor fits best with the target
fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharey=True)
for ax, col, label in zip(axes, ["vix", "vol_5d"], ["VIX", "5d vol"]):
    ax.scatter(data[col], data["fwd_vol_5d"], s=5, alpha=0.3)
    r = data[col].corr(data["fwd_vol_5d"])
    ax.set_title(f"{label} (correlation = {r:.2f})")
    ax.set_xlabel(f"{label} (%)")
    ax.grid(alpha=0.3)

axes[0].set_ylabel("Next Week vol")
plt.tight_layout()
plt.show()
```

**VIX**: 0.70. Vix lines up with the next weeks vol better than past volatility

**Past Week Volatility**: 0.61, still pretty strong but less than VIX

Stock Dataset for each listed stock


```python


TICKERS = ["SPY", "JNJ", "TSLA", "KO", "NVDA"]
START = "2010-07-01" # tesla started trading circa mid 2010s
END = "2022-12-31"
ANN = np.sqrt(252) * 100 # daily std annualized percentage (same units as VIX)

vix = yf.Ticker("^VIX").history(start=START)
vix.index = vix.index.tz_localize(None).normalize() # matching stock data dates
vix_close = vix["Close"]

def build_stock_data(ticker):
    """Predictors/targets for a stock. Rows are one trading day."""
    prices = yf.Ticker(ticker).history(start=START)["Close"]
    prices.index = prices.index.tz_localize(None).normalize()
    daily_ret = prices.pct_change()

    df = pd.DataFrame(index=prices.index)
    df["vol_5d"] = daily_ret.rolling(5).std() * ANN # Past Week
    df["vol_21d"] = daily_ret.rolling(21).std() * ANN # Past month
    df["vol_63d"] = daily_ret.rolling(63).std() * ANN # Past quarter
    df["VIX"] = vix_close

    df["fwd_vol"] = daily_ret.rolling(5).std().shift(-5) * ANN # Volatility over next 5 training dayss

    return df.dropna()

# Ticker Map

stock_data = {}
for t in TICKERS:
    stock_data[t] = build_stock_data(t)
    df = stock_data[t]
    print(f"{t}: {len(df)} rows, {df.index[0].date()} to {df.index[-1].date()}")




```

Next Week Volatility


```python
# Plot One: Next Week volatility over time for each stock

fig, axes = plt.subplots(len(TICKERS), 1, figsize=(13,2.6*len(TICKERS)),
                         sharex=True, sharey=True)

for ax, t in zip(axes, TICKERS):
    df = stock_data[t]
    ax.plot(df.index, df["fwd_vol"], lw=0.8, color="blue", label=f"{t} next-week vol")
    ax.plot(df.index, df["VIX"], lw=0.8, color="red",alpha=0.7, label="VIX")
    ax.set_ylabel("Annual Volatility (%)")
    ax.legend(loc="upper left")
    ax.grid(alpha=0.3)

axes[0].set_title("Next Week Volatility (Stock vs. VIX)")
plt.tight_layout()
plt.show()
```

SPY, JNJ(johnson n johnson), KO(coke) move with VIX. The blue lines high the red line pretty consistently and the spikes are prety similar and spike at similar tmes.

TSLA and NVDA are more volatile and spike more than VIX.. NVDA's spike in 2016 and 2023 don't show on VIX at all. Those are big earning surprises



Graph 2: Correlation of each predictor and next-week volatility

```python
corr_rows = []
for t in TICKERS:
    df = stock_data[t]
    corr_rows.append({
        "TICKER": t,
        "corr_VIX": df["VIX"].corr(df["fwd_vol"]),
        "corr_own_past_week": df["vol_5d"].corr(df["fwd_vol"])
    })
    table_corr = pd.DataFrame(corr_rows).set_index("TICKER").round(2)
    print(table_corr)

    table_corr.plot(kind="bar", figsize=(6,4), rot=0)
    plt.ylabel("Next Week Volatility Correlation")
    plt.title("Stocks that move with VIX")
    plt.ylim(0, 1)
    plt.grid(axis="y", alpha=0.3)
    plt.show()
```

Summary Stats

```python
summary_rows = []

for t in TICKERS:
    df = stock_data[t]
    train = df.loc[df.index <= END]
    test = df.loc[df.index > END]



    threshold = train["vol_5d"].quantile(0.75) # Volatile is top 25% of this specific stocks training weeks

    # 1 = high vol days, 0 = normal (saved into stock data for classification)
    df["volatile"] = (df["vol_5d"] > threshold).astype(int)
    summary_rows.append({
        "TICKER": t,
        "mean_vol %": df["vol_5d"].mean(),
        "median_vol %": df["vol_5d"].median(),
        "skew": df["vol_5d"].skew(),  # > 0 is a long right tail of spikes
        "threshold %": threshold,
        "pct_volatile_train": df.loc[train.index, "volatile"].mean() * 100,
        "pct_volatile_test": df.loc[test.index, "volatile"].mean() * 100
    })

summary_table = pd.DataFrame(summary_rows).set_index("TICKER").round(2)
print(summary_table)
```

SPY JNJ and KO average 13-14% annualized volatility while TSLA and NVDA are around 3x higher

SPY is quite stable compared to TSLA and NVDA


Every stock is skewed to the right. This means there are occasional large spikes in volatility.

SPY has the highest skew and TSLA the lowest. This shows TSLA is very volatile, its extremes are relative to their normal.

```python
# Now that we've tested it with the S&P 500, chip away at using every stock. 
# STOCKS picked NVDA, TSLA JNJ, KO, UNH, AMZN, HD, CAT, BA, AAPL, MSFT, JPM, GS
START = '2023-01-01'
END = '2026-01-01'

# Initializing the % change data frame
stocks = pd.DataFrame()
# NVDA data
nvda = yf.Ticker("NVDA").history(start=START, end=END)
nvda.index = nvda.index.tz_localize(None).normalize() # matching stock data dates
stocks["NVDA"] = (nvda["Close"] - nvda["Open"])/(nvda["Open"]) * 100

# TSLA data
tsla = yf.Ticker("TSLA").history(start=START, end=END)
tsla.index = tsla.index.tz_localize(None).normalize() # matching stock data dates
stocks["TSLA"] = (tsla["Close"] - tsla["Open"])/(tsla["Open"]) * 100

# JNJ data
jnj = yf.Ticker("JNJ").history(start=START, end=END)
jnj.index = jnj.index.tz_localize(None).normalize() # matching stock data dates
stocks["JNJ"] = (jnj["Close"] - jnj["Open"])/(jnj["Open"]) * 100

# KO data
ko = yf.Ticker("KO").history(start=START, end=END)
ko.index = ko.index.tz_localize(None).normalize() # matching stock data dates
stocks["KO"] = (ko["Close"] - ko["Open"])/(ko["Open"]) * 100

# UNH data
unh = yf.Ticker("UNH").history(start=START, end=END)
unh.index = unh.index.tz_localize(None).normalize() # matching stock data dates
stocks["UNH"] = (unh["Close"] - unh["Open"])/(unh["Open"]) * 100

# AMZN data
amzn = yf.Ticker("AMZN").history(start=START, end=END)
amzn.index = amzn.index.tz_localize(None).normalize() # matching stock data dates
stocks["AMZN"] = (amzn["Close"] - amzn["Open"])/(amzn["Open"]) * 100

# HD data
hd = yf.Ticker("HD").history(start=START, end=END)
hd.index = hd.index.tz_localize(None).normalize() # matching stock data dates
stocks["HD"] = (hd["Close"] - hd["Open"])/(hd["Open"]) * 100

# CAT data
cat = yf.Ticker("CAT").history(start=START, end=END)
cat.index = cat.index.tz_localize(None).normalize() # matching stock data dates
stocks["CAT"] = (cat["Close"] - cat["Open"])/(cat["Open"]) * 100

# BA data
ba = yf.Ticker("BA").history(start=START, end=END)
ba.index = ba.index.tz_localize(None).normalize() # matching stock data dates
stocks["BA"] = (ba["Close"] - ba["Open"])/(ba["Open"]) * 100

# AAPL data
aapl = yf.Ticker("AAPl").history(start=START, end=END)
aapl.index = aapl.index.tz_localize(None).normalize() # matching stock data dates
stocks["AAPL"] = (aapl["Close"] - aapl["Open"])/(aapl["Open"]) * 100

# MSFT data
msft = yf.Ticker("MSFT").history(start=START, end=END)
msft.index = msft.index.tz_localize(None).normalize() # matching stock data dates
stocks["MSFT"] = (msft["Close"] - msft["Open"])/(msft["Open"]) * 100

# JPM data
jpm = yf.Ticker("JPM").history(start=START, end=END)
jpm.index = jpm.index.tz_localize(None).normalize() # matching stock data dates
stocks["JPM"] = (jpm["Close"] - jpm["Open"])/(jpm["Open"]) * 100

# GS data
gs = yf.Ticker("GS").history(start=START, end=END)
gs.index = gs.index.tz_localize(None).normalize() # matching stock data dates
stocks["GS"] = (gs["Close"] - gs["Open"])/(gs["Open"]) * 100

stocks
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

vix = yf.Ticker("^VIX").history(start=START, end=END)
vix.index = vix.index.tz_localize(None).normalize() # matching stock data dates
vix_close = vix["Close"]

v = np.where(vix_close >= 22.612501, 1, 0)
#print(stocks.shape)

# This model "m" is the model for all 
m = LogisticRegression(random_state=0, max_iter=10000)
m.fit(stocks, v)
y_hat = m.predict(stocks)
C = confusion_matrix(v, y_hat)
ConfusionMatrixDisplay(C).plot();
print(f"Mean accuracy of LogReg using every stock (no train/test data): {round(m.score(stocks, v) * 100, 2)}%")
```

```python
# Using statsmodels to get another form of logistic regression that gives p-value summary. 
import statsmodels.api as sm

# Add an intercept column (statsmodels does not include one by default)
stocks_plus_const = sm.add_constant(stocks)

# Fit the logistic regression model
model = sm.Logit(v, stocks_plus_const)
result = model.fit()
result.summary()
```

```python
# Building df for testing high to low drop-offs.
START = '2023-01-01'
END = '2026-01-01'
TICKERS = ["NVDA", "TSLA", "JNJ", "KO", "UNH", "AMZN", "HD", "CAT", "BA", "AAPL", "MSFT", "JPM", "GS"]
hilo = pd.DataFrame()

for stock in TICKERS:
    S = yf.Ticker(stock).history(start=START, end=END)
    S.index = S.index.tz_localize(None).normalize() # matching stock data dates
    hilo[stock] = S["High"] - S["Low"]

vix = yf.Ticker("^VIX").history(start=START, end=END)
vix.index = vix.index.tz_localize(None).normalize() # matching stock data dates
vix_close = vix["Close"]

v = np.where(vix_close >= 22.612501, 1, 0)
#print(stocks.shape)

# This model "m" is the model for all 
m = LogisticRegression(random_state=0, max_iter=10000)
m.fit(hilo, v)
y_hat = m.predict(hilo)
C = confusion_matrix(v, y_hat)
ConfusionMatrixDisplay(C).plot();
print(f"Mean accuracy of LogReg using every stock (no train/test data): {round(m.score(hilo, v) * 100, 2)}%")
```

```python
# Add an intercept column (statsmodels does not include one by default)
hilo_plus_const = sm.add_constant(hilo)

# Fit the logistic regression model
model = sm.Logit(v, hilo_plus_const)
result = model.fit()
result.summary()
```

```python
# Testing the model for a single stock with the lowest p-value
ko = yf.Ticker("KO").history(start=START, end=END)
ko.index = ko.index.tz_localize(None).normalize() # matching stock data dates
khilo = np.array(ko["High"] - ko["Low"])
khilo = khilo.reshape(-1, 1)

m = LogisticRegression(random_state=0, max_iter=10000)
m.fit(khilo, v)
y_hat = m.predict(khilo)
C = confusion_matrix(v, y_hat)
ConfusionMatrixDisplay(C).plot();
print(f"Mean accuracy of LogReg using every stock (no train/test data): {round(m.score(khilo, v) * 100, 2)}%")
```
